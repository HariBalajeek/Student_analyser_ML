Edge Cases to Handle
1. Invalid Marks
Marks entered outside the valid range (e.g., -5 or 120).
Non-numeric input (e.g., "eighty" instead of 80).
2. Missing Data
Student record missing a subject score.
Empty student list (no data to analyze).
Missing name or roll number.
3. Duplicate Records
Two students with the same roll number.
Same student entered multiple times.
4. Boundary Grades
Averages exactly on grade cutoffs (e.g., 90 → A+, 75 → A).
Floating-point precision issues (e.g., 74.9999 should still be A).
5. Tie Situations
Multiple students with the same highest total marks (multiple toppers).
Subject-wise toppers with identical scores.
6. Empty or Invalid Subject List
No subjects defined.
Subjects with invalid names or duplicates.
7. Large Dataset
Hundreds or thousands of students — performance and memory usage.

8. Special Cases
Student with all zero marks (should still be processed).
Student with perfect marks (100 in all subjects).

Solution Design
1. Inputs
Student details: Name, Roll Number.
Marks: Dictionary of subjects and scores.
Multiple students stored in a list.
2. Processing Steps
Validate marks (numeric, 0–100).
Compute total marks per student.
Compute average marks.
Assign grades based on average.
Identify topper(s).
Compute class average.
Handle edge cases (missing data, ties, invalid marks).
3. Outputs
Individual student report (Name, Roll No, Total, Average, Grade).
Class-level analysis (Topper, Class Average, Subject-wise topper).
Summary report for all students.


```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks  # dictionary: {"Math": 85, "Science": 90, ...}

    def validate_marks(self):
        for subject, score in self.marks.items():
            if not isinstance(score, (int, float)) or score < 0 or score > 100:
                raise ValueError(f"Invalid mark {score} for {subject}")

    def total_marks(self):
        return sum(self.marks.values())

    def average_marks(self):
        return self.total_marks() / len(self.marks)

    def grade(self):
        avg = self.average_marks()
        if avg >= 90:
            return "A+"
        elif avg >= 75:
            return "A"
        elif avg >= 60:
            return "B"
        elif avg >= 50:
            return "C"
        else:
            return "F"

    def result_summary(self):
        return {
            "Name": self.name,
            "Roll No": self.roll_no,
            "Total": self.total_marks(),
            "Average": round(self.average_marks(), 2),
            "Grade": self.grade()
        }


class ResultAnalyzer:
    def __init__(self, students):
        self.students = students

    def class_average(self):
        total = sum(s.total_marks() for s in self.students)
        count = sum(len(s.marks) for s in self.students)
        return round(total / count, 2)

    def topper(self):
        max_marks = max(s.total_marks() for s in self.students)
        toppers = [s.name for s in self.students if s.total_marks() == max_marks]
        return toppers, max_marks

    def subject_wise_topper(self):
        subjects = self.students[0].marks.keys()
        toppers = {}
        for subject in subjects:
            max_score = max(s.marks[subject] for s in self.students)
            toppers[subject] = [s.name for s in self.students if s.marks[subject] == max_score]
        return toppers

    def generate_report(self):
        print("=== Student Result Analyzer ===")
        for s in self.students:
            print(s.result_summary())

        print("\nClass Average:", self.class_average())
        toppers, marks = self.topper()
        print("Topper(s):", ", ".join(toppers), "with", marks, "marks")

        print("\nSubject-wise Toppers:")
        for subject, names in self.subject_wise_topper().items():
            print(f"{subject}: {', '.join(names)}")


# Example usage
students = [
    Student("Hari", 1, {"Math": 85, "Science": 90, "English": 78}),
    Student("Anita", 2, {"Math": 92, "Science": 88, "English": 95}),
    Student("Ravi", 3, {"Math": 60, "Science": 55, "English": 70}),
]

analyzer = ResultAnalyzer(students)
analyzer.generate_report()
```
